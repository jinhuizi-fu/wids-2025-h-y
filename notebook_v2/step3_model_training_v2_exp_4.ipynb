```python
X_train = X_train.drop(columns=['participant_id'], axis=1)
y_train = y_train.drop(columns=['participant_id'], axis=1)
```

```python
# Custom scoring function for weighted F1 score
def weighted_f1(y_true, y_pred):
    # Ensure that y_true and y_pred are numpy arrays
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    
    f1_adhd = f1_score(y_true[:, 0], y_pred[:, 0])
    f1_sex = f1_score(y_true[:, 1], y_pred[:, 1])
    return (2/3) * f1_adhd + (1/3) * f1_sex
```

```python
def objective(trial):
    # Suggest hyperparameters
    num_units = trial.suggest_int('num_units', 32, 256)
    dropout_rate = trial.suggest_float('dropout_rate', 0.1, 0.5)
    learning_rate = trial.suggest_float('learning_rate', 1e-4, 1e-2, log=True)

    # Build the model
    input_layer = Input(shape=(X_train.shape[1],))
    shared = layers.Dense(num_units, activation='relu')(input_layer)
    shared = layers.Dropout(dropout_rate)(shared)
    shared = layers.Dense(num_units // 2, activation='relu')(shared)

    adhd_output = layers.Dense(1, activation='sigmoid', name='ADHD_Output')(shared)
    sex_output = layers.Dense(1, activation='sigmoid', name='Sex_Output')(shared)

    model = models.Model(inputs=input_layer, outputs=[adhd_output, sex_output])

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    # Train the model
    model.fit(
        X_train_scaled,
        {'ADHD_Output': y_train['ADHD_Outcome'], 'Sex_Output': y_train['Sex_F']},
        validation_data=(X_val_scaled, {'ADHD_Output': y_val['ADHD_Outcome'], 'Sex_Output': y_val['Sex_F']}),
        epochs=10,
        batch_size=32,
        verbose=0
    )

    # Predict on the validation set
    preds = model.predict(X_val_scaled)
    preds_adhd = (preds[0] > 0.5).astype(int)
    preds_sex = (preds[1] > 0.5).astype(int)

    # Calculate F1 scores
    f1_adhd = f1_score(y_val['ADHD_Outcome'], preds_adhd)
    f1_sex = f1_score(y_val['Sex_F'], preds_sex)

    # Weighted average of F1 scores
    final_score = (2/3) * f1_adhd + (1/3) * f1_sex

    return final_score
```

### tune model parameters

```python
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=50)

print("Best params:", study.best_trial.params)
```

```python
def multi_output_accuracy(y_true, y_pred):
    # Ensure y_true and y_pred are NumPy arrays
    y_true = np.array(y_true)
    y_pred = np.array(y_pred)
    # Compute accuracy for each target variable and return the mean
    return np.mean([accuracy_score(y_true[:, i], y_pred[:, i]) for i in range(y_true.shape[1])])
```
